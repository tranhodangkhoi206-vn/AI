```python
#Ham gan gia tri ban dau 
def GanGiaTriMau():
    MT = [[3,2,9],
          [4,1,7],
          [5,6,8]]
    return MT
```

```python
#Ham nhap ma tran 
def NhapMaTran(MT, N, M):
    for i in range(N):
        MT[i]=[0]*M
    for i in range(N):
        for i in range(N):
            for j in range(M):
                MT[i][j] = int(input("Nhap phan tu MT["+str(i)+"]["+str(j)+"]"))
```

```python
#Ham xuat ma tran 
def XuatMaTran(MT,M,N):
    for i in range(N):
        for j in range(M):
            print(MT[i][j], end = " ")
        print("\n")
```

```python
#Ham dem phan tu chan cot Y 
def DemPhanTuChanCotY(MT, N):
    Y = int(input("Dem so phan tu chan tren cot: "))
    Dem = 0
    for i in range(N):
        if (MT[i][Y-1]%2==0):
           Dem = Dem +1 
    print("So phan tu chan tren cot "+str(Y)+": ", Dem)
```

```python
#Ham tim phan tu lon nhat dong X 
def TimPTLonNhatDongX(MT, M):
    X= int(input("Tim phan tu lon nhat tren dong: "))
    Max = MT[X-1][0]
    for i in range(M):
       if (Max<MT[X-1][i]):
            Max=MT[X-1][i]
    print("Phan tu lon nhat dong "+str(X)+" la: ", Max)
```

```python
#Ham sap xep tang dan cot C
def SapXepTangDanCotC(MT, N):
    C= int(input("Sap xep theo gia tri tang dan tren cot: "))
    for i in range(N-1):
        for j in range(i+1, N):
            if(MT[i][C-1]>MT[j][C-1]):
                Tam=MT[i][C-1]
                MT[i][C-1]=MT[j][C-1]
                MT[j][C-1] = Tam
    for i in range (N):
        for j in range(M):
            print(MT[i][j], end= " ")
        print("\n")
```

```python
#Ham sap xep theo gia tri giam dan dong R
def SepXepGiamDanDongR(MT, M):
    R = int(input("Sap xep theo gia tri giam dan tren dong: "))
    for i in range(M-1):
        for j in range (i+1, M):
            if (MT[R-1][i]<MT[R-1][j]):
                Tam=MT[R-1][i]
                MT[M-1][i] = MT[R-1][j]
                MT[R-1][j] = Tam
    for  i in range(N):
        for j in range (M):
            print(MT[i][j],end = " ")
        print("\n")
```

```python
#Main
N = 3
M = 3
MT = GanGiaTriMau()
# NhapMaTran(MT, M, N)
XuatMaTran(MT, M, N)
DemPhanTuChanCotY(MT, N)
TimPTLonNhatDongX(MT, M)
SapXepTangDanCotC(MT, N)
SepXepGiamDanDongR(MT, M)
```
