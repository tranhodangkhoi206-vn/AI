```python
import math
```

```python
Sinhvien = [
    # Giờ học, Chuyên cần, Kết quả
    [1, 1, 1],  # SV1: Nhiều, Cao, Đỗ
    [0, 0, 0],  # SV2: Ít, Thấp, Trượt
    [1, 0, 1],  # SV3: Nhiều, Thấp, Đỗ
    [1, 0, 1],  # SV4: Nhiều, Thấp, Đỗ
    [0, 1, 1],  # SV5: Ít, Cao, Đỗ
    [0, 1, 1],  # SV6: Ít, Cao, Đỗ
    [0, 0, 0],  # SV7: Ít, Thấp, Trượt
    [1, 1, 1],  # SV8: Nhiều, Cao, Đỗ
    [0, 0, 0],  # SV9: Ít, Thấp, Trượt
    [1, 1, 1]   # SV10: Nhiều, Cao, Đỗ
]
```

```python
def DemSoLuongTrangThai(Sinhvien, trang_thai, cot_tt):
    dem = []
    for i in range(len(trang_thai)):
        dem.append(0)
    for i in Sinhvien:
        dem[i[cot_tt]] += 1
    return dem
```

```python
def XetTrangThai(Sinhvien, cot):
    trang_thai = []
    for i in Sinhvien:
        if i[cot] not in trang_thai:
            trang_thai.append(i[cot])
    trang_thai.sort()
    return trang_thai
```

```python
def XetViTri(Sinhvien, trang_thai, cot_tt):
    vi_tri =[] 
    for i in trang_thai:
        tam = []
        for j in range(len(Sinhvien)):
            if i == Sinhvien[j][cot_tt]:
                tam.append(j)
        vi_tri.append(tam)
    return vi_tri
```

```python
def TinhEntropyBanDau(Sinhvien, trang_thai_kq):
    h = 0
    dem = DemSoLuongTrangThai(Sinhvien, trang_thai_kq,-1)
    for i in range(len(trang_thai_kq)):
        p = dem[i] /len(Sinhvien) 
        h += -1 * p * math.log(p, 2)
    return h
```

```python
def TinhEntropyThuocTinh(Sinhvien, trang_thai_tt, dem, cot_tt, vi_tri):
    mang_h = []
    for l in trang_thai_tt:
        h = 0
        mang_dem = []
        for i in vi_tri:
            dem_tam = 0
            for j in i:
                if Sinhvien[j][cot_tt] == l:
                    dem_tam += 1
            mang_dem.append(dem_tam)
            p = mang_dem[vi_tri.index(i)] / dem[l]
            if p != 0:
                h += -1 * p * math.log(p, 2)
        mang_h.append(h)
    return mang_h
```

```python
def TinhInfomationGain(Sinhvien, dem, h):
    hs = 0
    for i in range(len(h)):
        hs += dem[i]/len(Sinhvien) * h[i]
    ig = TinhEntropyBanDau(Sinhvien, trang_thai_kq) - hs 
    return ig
```

```python
trang_thai_kq = XetTrangThai(Sinhvien, -1)
vi_tri = XetViTri(Sinhvien, trang_thai_kq, -1)
```

```python
def CayQuyetDinhID3(Sinhvien, vi_tri):
    if Sinhvien == []:
        return
    ig = []
    mang_tt_tt = []
    mang_h_tt = []
    Sinhvien_moi = []
    for i in range(len(Sinhvien[0]) - 1):
        trang_thai_tt = XetTrangThai(Sinhvien, i)
        mang_tt_tt.append(trang_thai_tt)
        dem_tt = DemSoLuongTrangThai(Sinhvien, trang_thai_tt, i)
        h_tt = TinhEntropyThuocTinh(Sinhvien, trang_thai_tt, dem_tt, i, vi_tri)
        mang_h_tt.append(h_tt)
        ig.append(TinhInfomationGain(Sinhvien, dem_tt, h_tt))
    max_ig = max(ig)
    vi_tri_tt = XetViTri(Sinhvien, mang_tt_tt[ig.index(max_ig)], ig.index(max_ig))
    for i in mang_h_tt[ig.index(max_ig)]:
        if i != 0.0:
            for j in vi_tri_tt[ig.index(max_ig)]:
                Sinhvien_moi.append(Sinhvien[j])
    i = 0
    while True:
        if i != ig.index(max_ig):
            vi_tri = XetViTri(Sinhvien_moi, mang_tt_tt[i], i)
            break
        i += 1
    print(Sinhvien)
    print(vi_tri_tt)
    return CayQuyetDinhID3(Sinhvien_moi, vi_tri)
```
