```python
def PhanGiai(m1, m2):
    # Tra ve danh sach (menh de giai thuc, literal bi khu trong m1)
    ket_qua = []
    for l in m1:
        if Doi(l) in m2:
            moi = (m1 - {l}) | (m2 - {Doi(l)})
            if not LaHangDung(moi):
                ket_qua.append((moi, l))
    return ket_qua
```

```python
def Robinson(ds, nguon):
    i = 0
    while i < len(ds):
        for j in range(i):
            for moi, l in PhanGiai(ds[j], ds[i]):
                if moi not in ds:
                    ds.append(moi)
                    nguon.append((j, i, l))
                    khu = l.replace("¬", "")
                    print(f"({len(ds)}) {InMenhDe(moi)}   "
                          f"[({j+1}) {InMenhDe(ds[j])}  +  ({i+1}) {InMenhDe(ds[i])}  | khu {khu}]")
                    if not moi:
                        return True
        i += 1
    return False
```

```python
def TimDuong(nguon, k, tap):
    # Di nguoc tu menh de k de lay cac dong tham gia chung minh
    if k in tap:
        return
    tap.add(k)
    if isinstance(nguon[k], tuple):
        j, i, _ = nguon[k]
        TimDuong(nguon, j, tap)
        TimDuong(nguon, i, tap)
```

```python
def InNguon(ds, nguon, k):
    if nguon[k] == "GT":
        return f"({k+1}) {InMenhDe(ds[k])}   [GT]"
    if nguon[k] == "KL":
        return f"({k+1}) {InMenhDe(ds[k])}   [phu dinh KL]"
    j, i, l = nguon[k]
    return (f"({k+1}) {InMenhDe(ds[k])}   "
            f"[({j+1}) + ({i+1}) | khu {l.replace('¬', '')}]")
```

```python
gt_input = "a->b,b->c"
kl_input = "a->c"

ds = []
nguon = []

# B1: dua gia thiet ve CNF
for gt in gt_input.split(","):
    for m in ChuyenThanhCNF(gt):
        if m not in ds:
            ds.append(m)
            nguon.append("GT")

# B2: phu dinh ket luan roi dua ve CNF
kl_phu_dinh = "¬(" + ChuyenThanhDangKyHieu(kl_input) + ")"
for m in ChuyenThanhCNF(kl_phu_dinh):
    if m not in ds:
        ds.append(m)
        nguon.append("KL")

print("B1+B2: Cac menh de ban dau")
for k in range(len(ds)):
    print(InNguon(ds, nguon, k))

print("\nB3: Ap dung quy tac giai")
if Robinson(ds, nguon):
    print("\n=> Ra menh de rong (□): ket luan DUNG")
    tap = set()
    TimDuong(nguon, len(ds) - 1, tap)
    print("\nDuong chung minh:")
    for k in sorted(tap):
        print(InNguon(ds, nguon, k))
else:
    print("\n=> Khong sinh them duoc menh de nao: ket luan KHONG suy ra duoc")
```
