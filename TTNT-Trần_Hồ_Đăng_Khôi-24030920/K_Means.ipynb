```python
import math
```

```python
# Sinhvien = [
#     [3, 8, 6, 7],
#     [5, 9, 7, 8],
#     [7, 10, 8, 9],
#     [4, 7, 5, 5],
#     [8, 9, 9, 10],
#     [6, 8, 7, 6],
# ]
Sinhvien = [
    [8, 2, 4, 0.5],
    [10, 3, 5, 0.6],
    [12, 4, 6, 0.67],
    [15, 5, 5, 1.0],
    [18, 6, 6, 1.0],
    [20, 7, 7, 1.0],
    [25, 10, 8, 1.25],
    [28, 12, 9, 1.33],
    [30, 14, 10, 1.4],
    [35, 16, 8, 2.0],
    [40, 20, 10, 2.0],
    [45, 24, 12, 2.0],
]
k = 3
soLuongThuocTinh = len(Sinhvien[0])
C = [Sinhvien[0], Sinhvien[4], Sinhvien[11]]
chia_cum = []
```

```python
def TinhKoangCach(sv, c):
    d = 0
    for i in range(len(sv)):
        d += (sv[i] - c[i]) * (sv[i] - c[i])
    d = math.sqrt(d)
    return d
```

```python
def CapNhatTamCum(sv, c, tam_cu):
    if c == []:
        return tam_cu 
    dem = 0
    tam_moi = []
    while dem < soLuongThuocTinh:
        x = 0
        for i in c:
            x += sv[i][dem]
        x /= len(c) 
        tam_moi.append(x)
        dem += 1
    return tam_moi
```

```python
def ThuatToanKMeans(sv, C,k, chia_cum):
    chia_cum_moi = []
    tam_cum = []
    for i in range(k):
        chia_cum_moi.append([])
    for i in range(len(sv)):
        do_dai = []
        for j in range(k):
            d = TinhKoangCach(sv[i], C[j])
            do_dai.append(d)
        d_min = min(do_dai)
        chia_cum_moi[do_dai.index(d_min)].append(i)
    if chia_cum != []:
        if chia_cum == chia_cum_moi:
            return chia_cum
    for i in range(k):
        tam_cum.append(CapNhatTamCum(sv, chia_cum_moi[i], C[i])) # Sau khi chia cum thi tinh toan tam cum moi
    return ThuatToanKMeans(sv, tam_cum,k, chia_cum_moi)
```

```python
array = ThuatToanKMeans(Sinhvien,C,k, [])
print(array)
```
