```python
def TimKiemASao(MT,D,Start,End):
    L = [Start]
    C = []
    Temp = {Start: 0}
    while True:
        if L == []:
            print("Khong tim thay dich") 
            return
        else:
            DinhDangXet =L.pop() 
            if DinhDangXet == End:
                print("Da tim toi dich")
                print(End,end="<-")
                while DinhDangXet != Start:
                    for i in range(D):
                        if MT[i][DinhDangXet] == -1:
                            print(i,end="<-")
                            DinhDangXet = i
            else:
                C.append(DinhDangXet)
                for i in range(D):
                    if MT[DinhDangXet][i] > 0 and i not in C:
                        HamDanhGia= MT[DinhDangXet][i] + H[i]
                        if i not in Temp or Temp[i] > HamDanhGia:
                            Temp[i] = HamDanhGia
                            if i not in L:
                                L.append(i)
                L = sorted(Temp)
```
