```python
MT = [
    #  A   B   C   D   E   F   G   H   I   K
    [  0, 12,  4, -1, -1, -1,  6, -1, -1, -1],  # A
    [ -1,  0, -1,  8, -1,  7, -1,  6, -1, -1],  # B
    [ -1, -1,  0,  5, 11,  2, -1, 17, -1, 27],  # C
    [  6, -1, -1,  0,  7, -1,  5, -1,  7, -1],  # D
    [ -1, -1, -1, -1,  0,  4, -1, -1,  9, -1],  # E
    [ -1, -1, -1, -1, -1,  0,  4, -1, -1, 19],  # F
    [ -1, -1, -1, -1, -1, -1,  0,  6, 11, -1],  # G
    [ -1, -1, -1, -1, -1, -1, -1,  0,  5, 15],  # H
    [ -1, -1, -1, -1, -1,  2, -1, -1,  0,  3],  # I
    [ -1, -1, -1, -1, -1, -1, -1, -1, -1,  0],  # K
]

H = {
    0: 16,  # A
    1: 12,  # B
    2: 15,  # C
    3: 9,   # D
    4: 10,  # E
    5: 14,  # F
    6: 11,  # G
    7: 8,   # H
    8: 2,   # I
    9: 0    # K
}

D = 10
Start = 0   # A
End = 9     # K
```

```python
def TimKiemASao(MT, D, Start, End):
    L = [Start]
    C = []
    L1 = {Start: H[Start]}  # Lưu giá trị đường đi
    while True:
        if L == []:
            print("Khong tim thay dich")
            return
        else:
            DinhDangXet = L.pop(0)
            if DinhDangXet == End:
                print("Da tim toi dich")
                print(DinhDangXet, end="<-")
                while DinhDangXet != Start:
                    for i in range(D):
                        if (
                            MT[i][DinhDangXet] > 0
                            and i in L1
                            and L1[i] - H[i] + MT[i][DinhDangXet] + H[DinhDangXet]
                            == L1[DinhDangXet]
                        ):
                            print(i, end="<-")
                            DinhDangXet = i
                            break
                print("BD")
                print("Với giá trị là :",L1[End])
                return
            else:
                C.append(DinhDangXet)
                for i in range(D):
                    if MT[DinhDangXet][i] > 0 and i not in C:
                        G = MT[DinhDangXet][i] + L1[DinhDangXet] - H[DinhDangXet]
                        HamDanhGia = G + H[i]
                        if i not in L1 or L1[i] > HamDanhGia:
                            L1[i] = HamDanhGia
                            if i not in L:
                                L.append(i)
                L1 = dict(sorted(L1.items(), key=lambda x: x[1]))
                L.sort(key=lambda x: L1[x])
```

```python
TimKiemASao(MT,D,Start,End)
```
