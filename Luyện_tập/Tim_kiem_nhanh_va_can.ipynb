```python
MT = [
    #  A   B   C   D   E   F   G   H   I   K
    [  0, 12,  4, -1, -1, -1,  6, -1, -1, -1],  # A
    [ -1,  0, -1,  8, -1,  7, -1,  6, -1, -1],  # B
    [ -1, -1,  0,  5, 11,  2, -1, 17, -1, 27],  # C
    [  6, -1, -1,  0,  7, -1,  5, -1,  7, -1],  # D
    [ -1, -1, -1, -1,  0,  4, -1, -1,  9, -1],  # E
    [ -1, -1, -1, -1, -1,  0,  4, -1, -1, 19],  # F
    [ -1, -1, -1, -1, -1, -1,  0,  6, 11, -1],  # G
    [ -1, -1, -1, -1, -1, -1, -1,  0,  5, 15],  # H
    [ -1, -1, -1, -1, -1,  2, -1, -1,  0,  3],  # I
    [ -1, -1, -1, -1, -1, -1, -1, -1, -1,  0],  # K
]

H = {
    0: 16,  # A
    1: 12,  # B
    2: 15,  # C
    3: 9,   # D
    4: 10,  # E
    5: 14,  # F
    6: 11,  # G
    7: 8,   # H
    8: 2,   # I
    9: 0    # K
}

D = 10
Start = 0   # A
End = 9     # K
```

```python
def TimKiemNhanhVaCan(MT, D, Start, End):
    L = [Start]
    C = []
    L1 = {Start: H[Start]}
    Dinhdangxet = 0
    Giatri = 0
    while True:
        if L == []:
            if Giatri == 0:
                print("Khong tim thay dich")
                return
            else:
                Dinhdangxet = End
                print("Da tim thay dich")
                print(Dinhdangxet, end="<-")
                while Dinhdangxet != Start:
                    for i in range(D):
                        if (
                            MT[i][Dinhdangxet] > 0
                            and i in L1
                            and L1[i] - H[i] + MT[i][Dinhdangxet] + H[Dinhdangxet]
                            == L1[Dinhdangxet]
                        ):
                            print(i, end="<-")
                            Dinhdangxet = i
                            break
                print("BD")
                print("Voi chi phi la:", Giatri)
                return
        else:
            Dinhdangxet = L.pop(0)
            if Dinhdangxet == End:
                Giatri = L1[End]
            Temp = []
            C.append(Dinhdangxet)
            for i in range(D):
                if MT[Dinhdangxet][i] > 0:
                    G = L1[Dinhdangxet] + MT[Dinhdangxet][i] - H[Dinhdangxet]
                    Hamdanhgia = G + H[i]
                    if Giatri != 0 and Hamdanhgia >= Giatri:
                        continue
                    if i not in L1 or L1[i] > Hamdanhgia:
                        L1[i] = Hamdanhgia
                        if i not in L:
                            Temp.append(i)
                        if i in C:
                            C.remove(i)
            L1 = dict(sorted(L1.items(), key=lambda x: x[1]))
            Temp.sort(key=lambda x: L1[x], reverse=True)
            for i in Temp:
                L.insert(0, i)
            print(L1)
            print(L)
```

```python
TimKiemNhanhVaCan(MT, D, Start, End)
```
